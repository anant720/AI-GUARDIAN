```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set aesthetic style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

## 📂 Data Loading
We ingest the `comparison_results.json` generated by the backend's benchmarking engine.

```python
with open('../backend/comparison_results.json') as f:
    data = json.load(f)

summary = data['summary']
breakdown = data['breakdown']
raw_results = data['raw_results']

print(f"Data Loaded! Total Scenarios: {summary['total']}")
```

## 1. Overall Accuracy Comparison
The core metric: How does AI Guardian compare with Industry Baselines (Google Safe Browsing & Truecaller)?

```python
metrics = ['Overall Accuracy', 'Scam Catch Rate (Recall)']
guardian_vals = [summary['guardian_overall_accuracy'], summary['guardian_catch_rate']]
industry_vals = [summary['industry_overall_accuracy'], summary['industry_catch_rate']]

x = np.arange(len(metrics))
width = 0.35

fig, ax = plt.subplots()
rects1 = ax.bar(x - width/2, guardian_vals, width, label='AI Guardian v3.0.0', color='#007bff')
rects2 = ax.bar(x + width/2, industry_vals, width, label='Industry Baseline', color='#ff4b5c')

ax.set_ylabel('Scores (0-1)')
ax.set_title('AI Guardian vs Industry Performance')
ax.set_xticks(x)
ax.set_xticklabels(metrics)
ax.legend()

plt.show()
```

## 2. Category Performance (The Deep-Dive)
We broken down detection rates into critical scam categories: **Banking**, **Tax**, **Crypto**, and **Indian SMS Scams**.

```python
categories = [cat for cat in breakdown.keys() if cat != 'safe']
g_cat = [breakdown[cat]['guardian'] for cat in categories]
i_cat = [breakdown[cat]['industry'] for cat in categories]

df_cat = pd.DataFrame({
    'Category': categories * 2,
    'Accuracy': g_cat + i_cat,
    'Provider': ['AI Guardian'] * len(categories) + ['Industry Baseline'] * len(categories)
})

sns.barplot(data=df_cat, x='Category', y='Accuracy', hue='Provider', palette=['#007bff', '#ff4b5c'])
plt.title('Accuracy by Scam Category')
plt.ylim(0, 1.0)
plt.show()
```

## 3. Latency Distribution
AI Guardian uses a 10-layer pipeline. How fast is it in reality?

```python
latencies = [r['guardian']['latency_ms'] for r in raw_results if 'guardian' in r]
plt.hist(latencies, bins=15, color='#28a745', alpha=0.7)
plt.axvline(np.mean(latencies), color='red', linestyle='dashed', linewidth=1, label=f"Mean: {np.mean(latencies):.1f}ms")
plt.title('AI Guardian Pipeline Latency (ms)')
plt.xlabel('Milliseconds')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
